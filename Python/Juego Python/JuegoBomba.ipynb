```python
import tkinter as tk
from tkinter import messagebox
import time

# Código correcto y tiempo límite en segundos
codigo_correcto = "44322S"

tiempo_limite = 900 

def verificar_codigo():
    global tiempo_restante
    codigo_ingresado = entry_codigo.get()
    if codigo_ingresado == codigo_correcto:
        messagebox.showinfo("¡Bieeeeeeeeen! ¡GANASTE!","¡Felicidades, el código es correcto!!!! :D")
        ventana.destroy()
    else:
        messagebox.showerror("Código Incorrecto", "El código ingresado no es correcto. Inténtalo de nuevo")

def actualizar_temporizador():
    global tiempo_restante
    if tiempo_restante > 0: # se llama cada segundo
        tiempo_restante -= 1
        label_tiempo.config(text=f"Tiempo: {tiempo_restante} segundos")
        ventana.after(1000, actualizar_temporizador)
    else:
        messagebox.showerror("¡OOOOOOOOOHhhhhhhhh Nooooooooooo PERDISTE!", "¡Se acabó el tiempo :(")
        ventana.destroy()

ventana = tk.Tk()
ventana.title("JUEGO BOMBA")
ventana.geometry("400x250")
ventana.configure(bg="black")
ventana.iconbitmap("bomba.ico")

# Etiqueta para mostrar el tiempo restante
tiempo_restante = tiempo_limite
label_tiempo = tk.Label(ventana, text=f"Tiempo: {tiempo_restante} segundos", font=("Arial", 14),bg="black",fg="yellow")
label_tiempo.pack(pady=20)

# Etiqueta para introducir el código
label_codigo = tk.Label(ventana, text="Introduce el código de 6 caracteres:", font=("Arial", 12),bg="black",fg="red")
label_codigo.pack()

# Campo de texto para ingresar el código
entry_codigo = tk.Entry(ventana, font=("Arial", 12), width=20,bg="grey")
entry_codigo.pack(pady=10)

# Botón para verificar el código
boton_verificar = tk.Button(ventana, text="Verificar", font=("Arial", 12), command=verificar_codigo,bg="red",fg="black")
boton_verificar.pack()

# Iniciar el temporizador
actualizar_temporizador()

# Iniciar el bucle de la interfaz gráfica
ventana.mainloop()
```
