#### 3.- Dada la cadena _texto_, crea un diccionario donde la clave sea cada letra y el valor sea la cantidad de veces que aparece. Convierte el diccionario en una lista de tuplas (letra, frecuencia) ordenada de mayor a menor frecuencia. 

```python
texto = "banana"
```

```python
frecuencias = {} #Creo un diccionario sin respuestas al que ir sumándole los resultados obtenidos

for letra in texto:
    if letra in frecuencias:
        frecuencias[letra] += 1
    else:
        frecuencias[letra] = 1
#Si la letra ya está introducida en el diccionario se le suma 1 y si no se pone uno para empezar a sumar si vuelve a aparecer

lista_frecuencias = list(frecuencias.items())
#Obtenemos los pares clave-valor para cada letra-valor

lista_frecuencias.sort(key=lambda x: x[1], reverse=True)
#"lambda x: x[1]" como key para decirle a sort() que ordende las tuplas por el segundo elemento, por la frecuencia de cada letra

print("Diccionario:", frecuencias)
print("Lista ordenada:", lista_frecuencias)
```

#### 6.- Dado el string _frase_, obtén un set con todas las palabras únicas. Crea un diccionario que cuente cuántas veces aparece cada palabra. Devuelve el diccionario pero solo con las palabras que aparezcan más de 1 vez.

```python
frase = "En el camino, uno se da cuenta de que en un coche, uno siempre está en un compartimento, y, porque está acostumbrado, no se da cuenta de que, a través de esa ventana, todo lo que ve es simplemente más televisión"
```

```python
frase_limpia = frase.lower().replace(",", "").replace(".", "")
#Ponemos la frase en minúscula y sustituimos las comas y puntos por espacios

palabras = frase_limpia.split()
#Cortamos la frase por cada espacio

palabras_unicas = set(palabras)
#Creamos un set para registrar cada palabra y, al ser un set, no se repetirán

frecuencias = {}
#Creamos un diccionario para registrar cada palabra con un contador

for palabra in palabras:
    if palabra in frecuencias:
        frecuencias[palabra] += 1
    else:
        frecuencias[palabra] = 1

frecuencias_repetidas = {
    palabra: cantidad
    for palabra, cantidad in frecuencias.items()
    if cantidad > 1
}
#Si la palabra está repetida entonces se guarda y se almacena con su segundo valor

print("Palabras únicas:", palabras_unicas)
print()
print("Frecuencias de palabras repetidas:", frecuencias_repetidas)
```

#### 10.- Dado el dicionario _datos_, pon la primera letra de "nombre" en mayúsculas. Calcula la suma total de "puntos" y guarda el total en una nueva clave "total". Marca un booleano como "aprobado" si el total es mayor a 50. Finalmente, crea un set con todos los nombres de los usuarios aprobados. 

```python
datos = [
    {"id":1,"nombre":"ana","puntos":(10,20,30)},
    {"id":2,"nombre":"luis","puntos":(5,15)},
    {"id":3,"nombre":"marta","puntos":(100,50,25)}
]
```

```python
aprobados = set()
#set para los nombres

for usuario in datos:
    usuario["nombre"] = usuario["nombre"].capitalize()

    total = sum(usuario["puntos"])
    usuario["total"] = total

    usuario["aprobado"] = total > 50
     # Comprobar si está aprobado
    
    if usuario["aprobado"]:
        aprobados.add(usuario["nombre"])
        # Añadir el nombre al set si está aprobado

for usuario in datos:
    print(
        f"ID: {usuario['id']} | "
        f"Nombre: {usuario['nombre']} | "
        f"Puntos: {usuario['puntos']} | "
        f"Total: {usuario['total']} | "
        f"Aprobado: {usuario['aprobado']}"
    ) 

print()
print("Usuarios aprobados:", aprobados)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python
resultado = {}
#set para los resultados

for estudiante in estudiantes.values():
    nombre = estudiante["nombre"].upper()
    notas = estudiante["notas"]

    promedio = sum(notas) / len(notas)
    aprobado = promedio >= 20
    max_min = (max(notas), min(notas))

    resultado[nombre] = {
        "promedio": promedio,
        "aprobado": aprobado,
        "max_min": max_min
    }

for nombre, datos in resultado.items():
    print(f"Estudiante: {nombre}")
    print(f"Promedio: {datos['promedio']}")
    print(f"Aprobado: {datos['aprobado']}")
    print(f"Nota máxima y mínima: {datos['max_min']}")
    print("-" * 30)
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
import json

with open("notas_alumnos.txt", "w", encoding="utf-8") as archivo: #("Nombre del archivo de texto", "modo escritura", "El modo utf-8 permite guardar caracteres especiales")
    archivo.write("Ana,10,20,30\n") #.write escribe texto en el archivo
    archivo.write("Luis,5,15\n")
    archivo.write("Marta,100,50,25\n")

resultado = []
#Lista vacía para hacer un diccionario por alumno

with open("notas_alumnos.txt", "r", encoding="utf-8") as archivo: #"r" significa modo lectura
    for linea in archivo:
        datos = linea.strip().split(",")

        nombre = datos[0].capitalize()
        notas = [int(nota) for nota in datos[1:]] #Convierte cada elemento de cada nota de str a int, excepto el primero

        promedio = sum(notas) / len(notas)
        aprobado = promedio >= 20

        alumno = {
            "nombre": nombre,
            "promedio": promedio,
            "aprobado": aprobado
        }

        resultado.append(alumno) #Añade cada alumno al diccionario
        print(alumno)

with open("resultado.json", "w", encoding="utf-8") as archivo:
    json.dump(resultado, archivo, ensure_ascii=False, indent=4)
    #json.dump() convierte la lista a formato json y escribe en el archivo
    #ensure_ascii=False permite conservar "ñ" o "vocales acentuadas"
    #indent=4 organiza con una indentación de 4 espacios para facilitar la lectura
```

#### 14.- Crea el archivo _usuarios.json_con el contenido del diccionario _datos_. Abre el archivo de nuevo y convierte todos los nombres a mayúsculas, luego crea un archivo _usuarios.csv_ con formato:
- id,nombre,roles
- 1,ANA,user|admin
- 2,LUIS,user
- 3,MARTA,moderator|user

```python
import json
import csv

datos = {
    "1": {"nombre": "ana", "roles": {"user", "admin"}},
    "2": {"nombre": "luis", "roles": {"user"}},
    "3": {"nombre": "marta", "roles": {"moderator", "user"}}
}

#Convertir los sets de roles en listas y guardar el JSON
datos_json = {}

for id_usuario, usuario in datos.items():
    datos_json[id_usuario] = {
        "nombre": usuario["nombre"],
        "roles": sorted(list(usuario["roles"]))
    }

with open("usuarios.json", "w", encoding="utf-8") as archivo:
    json.dump(datos_json, archivo, ensure_ascii=False, indent=4)
    #Guardar el diccionario en usuarios.json

with open("usuarios.json", "r", encoding="utf-8") as archivo:
    usuarios = json.load(archivo)
    #Lee el archivo

for usuario in usuarios.values():
    usuario["nombre"] = usuario["nombre"].upper()

with open("usuarios.csv", "w", encoding="utf-8", newline="") as archivo:
    escritor = csv.writer(archivo)
#archivo csv
    
    escritor.writerow(["id", "nombre", "roles"])

    for id_usuario, usuario in usuarios.items():
        roles = "|".join(usuario["roles"])
        escritor.writerow([id_usuario, usuario["nombre"], roles])
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes. 

```python
import json
import string

# Leer el archivo
with open("fragmento-libro.txt", "r", encoding="utf-8") as archivo:
    texto = archivo.read()

# Normalizar el texto: minúsculas y sin signos
signos = string.punctuation + "¡¿«»“”‘’" #Reune los signos a eliminar
texto = texto.lower().translate(str.maketrans("", "", signos)) #Crea una tabla para eliminar los signos

# Separar las palabras
palabras = texto.split()

# Contar la frecuencia de cada palabra
frecuencias = {}
for palabra in palabras:
    if palabra in frecuencias:
        frecuencias[palabra] += 1
    else:
        frecuencias[palabra] = 1

# Seleccionar las 5 palabras más comunes
cinco_mas_comunes = sorted(
    frecuencias.items(), #Obtiene pares (palabra, frecuencia)
    key=lambda elemento: elemento[1], #Ordena por el elemento en la posición 1 del par 
    reverse=True
)[:5] #Selecciona los primeros 5 elementos

# Guardar el resultado en JSON
with open("frecuencias.json", "w", encoding="utf-8") as archivo:
    json.dump(cinco_mas_comunes, archivo, ensure_ascii=False, indent=4)

print(cinco_mas_comunes)
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
import csv

for nombre_archivo in ["ventas1.csv", "ventas2.csv", "ventas3.csv"]:
    with open(nombre_archivo, "w", newline="", encoding="utf-8") as archivo:
        escritor = csv.writer(archivo)
        escritor.writerow(["producto", "cantidad"])
        escritor.writerow(["A", 10])
        escritor.writerow(["B", 5])

totales = {}

# Leer los tres archivos y sumar las cantidades
for nombre_archivo in ["ventas1.csv", "ventas2.csv", "ventas3.csv"]:
    with open(nombre_archivo, "r", newline="", encoding="utf-8") as archivo:
        lector = csv.DictReader(archivo) #Lee cada fila como un diccionario usando los nombres de las columnas

        for fila in lector:
            producto = fila["producto"] #Obtiene el producto
            cantidad = int(fila["cantidad"]) #Convierte la cantidad del texto a número entero

            if producto in totales: #Contador
                totales[producto] += cantidad
            else:
                totales[producto] = cantidad

# Guardar los totales en un único archivo CSV
with open("total_ventas.csv", "w", newline="", encoding="utf-8") as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(["producto", "cantidad"])

    for producto, cantidad in totales.items():
        escritor.writerow([producto, cantidad])

print(totales)
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
import json

alumnos = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]}
]

with open("alumnos.json", "w", encoding="utf-8") as archivo:
    json.dump(alumnos, archivo, ensure_ascii=False, indent=4)

with open("alumnos.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

# Crear el diccionario con los resultados
resultados = {}

for alumno in datos:
    nombre = alumno["nombre"].capitalize()
    notas = alumno["notas"]

    resultados[nombre] = {
        "promedio": sum(notas) / len(notas),
        "max": max(notas),
        "min": min(notas),
        "aprobado": sum(notas) / len(notas) >= 20
    }

for nombre, datos in resultados.items():
    print(nombre, datos)
    print()
```
