## TASK 1: Get a `COVID-19 pandemic` Wiki page using HTTP request


First, let's write a function to use HTTP request to get a public COVID-19 Wiki page.

Before you write the function, you can open this public page from this 

URL https://en.wikipedia.org/w/index.php?title=Template:COVID-19_testing_by_country using a web browser.

The goal of task 1 is to get the html page using HTTP request (`httr` library)


```r

    
  # Our target COVID-19 wiki page URL is: https://en.wikipedia.org/w/index.php?title=Template:COVID-19_testing_by_country  
  # Which has two parts: 
    # 1) base URL `https://en.wikipedia.org/w/index.php  
    # 2) URL parameter: `title=Template:COVID-19_testing_by_country`, seperated by question mark ?
    
  # Wiki page base
  wiki_base_url <- "https://en.wikipedia.org/w/index.php"
  # You will need to create a List which has an element called `title` to specify which page you want to get from Wiki
  # in our case, it will be `Template:COVID-19_testing_by_country`
 
  # - Use the `GET` function in httr library with a `url` argument and a `query` arugment to get a HTTP response
    
  # Use the `return` function to return the response


library(httr)

get_wiki_covid19_page <- function() {

  # Wiki page base URL
  wiki_base_url <- "https://en.wikipedia.org/w/index.php"

  # Create a list of query parameters
  query_params <- list(
    title = "Template:COVID-19_testing_by_country"
  )

  # Send the HTTP GET request
  response <- GET(
    url = wiki_base_url,
    query = query_params
  )

  # Return the HTTP response
  return(response)
}
# Get the HTTP response
response <- get_wiki_covid19_page()

# Display the HTTP status code
status_code(response)

# Display the first 500 characters of the returned HTML
page_html <- content(response, as = "text", encoding = "UTF-8")
cat(substr(page_html, 1, 500))

# Read the exported CSV into a new data frame
covid_data_from_csv <- read.csv("covid19_data.csv")

# Select rows 5 to 10 and the requested columns
covid_subset <- covid_data_from_csv[
  5:10,
  c("country", "confirmed")
]

# Display the subset
print(covid_subset)

# Call the function
response <- get_wiki_covid19_page()
```

## TASK 3: Pre-process and export the extracted data frame

The goal of task 3 is to pre-process the extracted data frame from the previous step, and export it as a csv file


Let's get a summary of the data frame


```r
# Print the summary of the data frame
# Get a summary of the data frame
summary(covid_table)
str(covid_table)
dim(covid_table)
head(covid_table)
```

As you can see from the summary, the columns names are little bit different to understand and some column data types are not correct. For example, the `Tested` column shows as `character`. 

As such, the data frame read from HTML table will need some pre-processing such as removing irrelvant columns, renaming columns, and convert columns into proper data types.


We have prepared a pre-processing function for you to conver the data frame but you can also try to write one by yourself


```r
library(dplyr)

preprocess_covid_data <- function(df) {
  df <- df %>%
    # Remove irrelevant columns (adjust indices as needed)
    select(-c(1, 3)) %>%

    # Rename columns
    rename(
      Country = 1,
      Tested = 2,
      Confirmed = 3,
      Deaths = 4,
      Tests_per_million = 5
    ) %>%

    # Convert numeric columns
    mutate(
      Tested = as.numeric(gsub(",", "", Tested)),
      Confirmed = as.numeric(gsub(",", "", Confirmed)),
      Deaths = as.numeric(gsub(",", "", Deaths)),
      Tests_per_million = as.numeric(gsub(",", "", Tests_per_million))
    )

  return(df)
}

# Apply preprocessing
covid_data <- preprocess_covid_data(covid_table)

# Review the cleaned data
summary(covid_data)
str(covid_data)
```

```r
reprocess_covid_data_frame <- function(data_frame) {

  # Remove the World row
  data_frame <- data_frame[
    data_frame$Country.or.region != "World", ]

  # Remove the last row, if present
  data_frame <- data_frame[1:min(172, nrow(data_frame)), ]

  # Remove unnecessary columns
  data_frame$Ref. <- NULL
  data_frame$Units.b. <- NULL

  # Rename columns
  names(data_frame) <- c(
    "country",
    "date",
    "tested",
    "confirmed",
    "confirmed.tested.ratio",
    "tested.population.ratio",
    "confirmed.population.ratio"
  )

  # Convert column data types
  data_frame$country <- as.factor(data_frame$country)
  data_frame$date <- as.factor(data_frame$date)

  numeric_columns <- c(
    "tested",
    "confirmed",
    "confirmed.tested.ratio",
    "tested.population.ratio",
    "confirmed.population.ratio"
  )

  data_frame[numeric_columns] <- lapply(
    data_frame[numeric_columns],
    function(x) as.numeric(gsub(",", "", x))
  )

  return(data_frame)
}

# Apply the preprocessing function
covid_data <- reprocess_covid_data_frame(covid_table)

# Inspect the cleaned data
summary(covid_data)
str(covid_data)
```
